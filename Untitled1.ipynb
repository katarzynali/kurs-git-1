```python
class Calculator:
  x: int
  y: int

  def __init__(self, x:int, y: int) -> None:
    self._x = x
    self._y = y
  def sum(self) -> int:
    return self._x + self._y
  def multiply(self) -> int:
    return self._x * self._y
  def substract(self) -> int:
    return self._x - self._y
  def division(self) -> float:
    return self._x / self._y


calc = Calculator(12, 4)
print(f"Dane: x={calc._x}, y={calc._y}")
print("-" * 30)

print(f"Dodawanie (12 + 4): {calc.sum()}")
print(f"Mnożenie (12 * 4): {calc.multiply()}")
print(f"Odejmowanie (12 - 4): {calc.substract()}")
print(f"Dzielenie (12 / 4): {calc.division()}")
```

```python
class BrakSrodkowError(Exception):
    def __init__(self, saldo, kwota_wyplaty):
        self.saldo = saldo
        self.kwota_wyplaty = kwota_wyplaty
        super().__init__(f"Błąd: Niewystarczające środki. Saldo: {saldo:.2f} zł. Żądana kwota: {kwota_wyplaty:.2f} zł.")

class NieprawidlowaKwotaError(Exception):
    def __init__(self, kwota):
        self.kwota = kwota
        super().__init__(f"Błąd: Kwota musi być dodatnia. Podano: {kwota:.2f} zł.")

class Bankomat:

    def __init__(self, poczatkowe_saldo=0.0):
        if poczatkowe_saldo < 0:
            poczatkowe_saldo = 0.0 # Upewniamy się, że saldo nie jest ujemne na starcie
        self.saldo = poczatkowe_saldo
        print(f"Bankomat zainicjowany. Początkowe saldo: {self.saldo:.2f} zł.")

    def sprawdz_saldo(self):
        print(f" Sprawdzenie Salda ")
        print(f"Aktualne saldo: {self.saldo:.2f} zł.")
        return self.saldo

    def wplata(self, kwota):
        if kwota <= 0:
            raise NieprawidlowaKwotaError(kwota)

        self.saldo += kwota
        print(f" Wpłata ")
        print(f"Pomyślnie wpłacono: {kwota:.2f} zł.")
        print(f"Nowe saldo: {self.saldo:.2f} zł.")

    def wyplata(self, kwota):

        if kwota <= 0:
            raise NieprawidlowaKwotaError(kwota)

        if kwota > self.saldo:
            raise BrakSrodkowError(self.saldo, kwota)

        self.saldo -= kwota
        print(f" Wypłata ")
        print(f"Pomyślnie wypłacono: {kwota:.2f} zł.")
        print(f"Nowe saldo: {self.saldo:.2f} zł.")

moj_bankomat = Bankomat(poczatkowe_saldo=1000.00)

print("Pierwszy: Poprawne operacje ")
try:
    moj_bankomat.wplata(500.00)
    moj_bankomat.sprawdz_saldo()
    moj_bankomat.wyplata(200.00)
except (NieprawidlowaKwotaError, BrakSrodkowError) as e:
    print(f"Wystąpił błąd: {e}")

print("Drugi: Błąd 'Brak Środków' ")
try:

    moj_bankomat.wyplata(2000.00)
except BrakSrodkowError as e:
    print(f"Wyjątek Złapany (BrakSrodkowError) !")
    print(f"KOMUNIKAT: {e}")
    moj_bankomat.sprawdz_saldo()


print("Trzeci: Błąd 'Nieprawidłowa Kwota' ")
try:
    moj_bankomat.wyplata(-50.00)
except NieprawidlowaKwotaError as e:
    print(f"Wyjątek Złapany (NieprawidlowaKwotaError) !")
    print(f"KOMUNIKAT: {e}")

print("Czwarte: Błąd 'ValueError' (nie-liczba) ")
try:
    kwota_uzytkownika = "sto złotych"
    float(kwota_uzytkownika)
    moj_bankomat.wplata(float(kwota_uzytkownika))
except ValueError:
    print("Wyjątek Złapany (ValueError) !")
    print("Błąd: Wprowadzona wartość nie jest poprawną liczbą.")

moj_bankomat.sprawdz_saldo()
```

```python
class Macierz:

    def stworz_macierz_2x2():
        print("Wprowadzanie macierzy 2x2 ")

        try:
            w1_e1 = int(input("Podaj element (1, 1): "))
            w1_e2 = int(input("Podaj element (1, 2): "))
            w2_e1 = int(input("Podaj element (2, 1): "))
            w2_e2 = int(input("Podaj element (2, 2): "))
        except ValueError as e:

            print(f"FATALNY BŁĄD: Wprowadzono niepoprawną wartość! {e}")
            return [[0, 0], [0, 0]]

        macierz = [
            [w1_e1, w1_e2],
            [w2_e1, w2_e2]
        ]
        return macierz

    def wyswietl_macierz(macierz):
        print(f"| {macierz[0][0]} {macierz[0][1]} |")
        print(f"| {macierz[1][0]} {macierz[1][1]} |")

    def dodaj_macierze(macierz_a, macierz_b):
        print("Dodawanie macierzy")
        wynik = [
            [macierz_a[0][0] + macierz_b[0][0], macierz_a[0][1] + macierz_b[0][1]],
            [macierz_a[1][0] + macierz_b[1][0], macierz_a[1][1] + macierz_b[1][1]]
        ]
        return wynik

    def odejmij_macierze(macierz_a, macierz_b):
        print("Odejmowanie macierzy")
        wynik = [
            [macierz_a[0][0] - macierz_b[0][0], macierz_a[0][1] - macierz_b[0][1]],
            [macierz_a[1][0] - macierz_b[1][0], macierz_a[1][1] - macierz_b[1][1]]
        ]
        return wynik


macierz1 = Macierz.stworz_macierz_2x2()
print("Macierz 1:")
Macierz.wyswietl_macierz(macierz1)

macierz2 = Macierz.stworz_macierz_2x2()
print("Macierz 2:")
Macierz.wyswietl_macierz(macierz2)

suma = Macierz.dodaj_macierze(macierz1, macierz2)
print("Wynik dodawania:")
Macierz.wyswietl_macierz(suma)

roznica = Macierz.odejmij_macierze(macierz1, macierz2)
print("Wynik odejmowania:")
Macierz.wyswietl_macierz(roznica)
```
