```python
class Calculator:

  def __init__(self, x:int, y: int) -> None:
    self._x = x
    self._y = y
  def sum(self) -> int:
    return self._x + self._y
  def multiply(self) -> int:
    return self._x * self._y
  def substract(self) -> int:
    return self._x - self._y
  def division(self) -> float:
    return self._x / self._y


calc = Calculator(12, 4)
print(f"Dane: x={calc._x}, y={calc._y}")
print("-" * 30)

print(f"Dodawanie (12 + 4): {calc.sum()}")
print(f"Mnożenie (12 * 4): {calc.multiply()}")
print(f"Odejmowanie (12 - 4): {calc.substract()}")
print(f"Dzielenie (12 / 4): {calc.division()}")
```

```python
class BrakSrodkowError(Exception):
    def __init__(self, saldo, kwota_wyplaty):
        self.saldo = saldo
        self.kwota_wyplaty = kwota_wyplaty
        super().__init__(f"Błąd: Niewystarczające środki. Saldo: {saldo:.2f} zł. Żądana kwota: {kwota_wyplaty:.2f} zł.")

class NieprawidlowaKwotaError(Exception):
    def __init__(self, kwota):
        self.kwota = kwota
        super().__init__(f"Błąd: Kwota musi być dodatnia. Podano: {kwota:.2f} zł.")

class Bankomat:

    def __init__(self, poczatkowe_saldo=0.0):
        if poczatkowe_saldo < 0:
            poczatkowe_saldo = 0.0 # Upewniamy się, że saldo nie jest ujemne na starcie
        self.saldo = poczatkowe_saldo
        print(f"Bankomat zainicjowany. Początkowe saldo: {self.saldo:.2f} zł.")

    def sprawdz_saldo(self):
        print(f" Sprawdzenie Salda ")
        print(f"Aktualne saldo: {self.saldo:.2f} zł.")
        return self.saldo

    def wplata(self, kwota):
        if kwota <= 0:
            raise NieprawidlowaKwotaError(kwota)

        self.saldo += kwota
        print(f" Wpłata ")
        print(f"Pomyślnie wpłacono: {kwota:.2f} zł.")
        print(f"Nowe saldo: {self.saldo:.2f} zł.")

    def wyplata(self, kwota):

        if kwota <= 0:
            raise NieprawidlowaKwotaError(kwota)

        if kwota > self.saldo:
            raise BrakSrodkowError(self.saldo, kwota)

        self.saldo -= kwota
        print(f" Wypłata ")
        print(f"Pomyślnie wypłacono: {kwota:.2f} zł.")
        print(f"Nowe saldo: {self.saldo:.2f} zł.")

moj_bankomat = Bankomat(poczatkowe_saldo=1000.00)

print("Pierwszy: Poprawne operacje ")
try:
    moj_bankomat.wplata(500.00)
    moj_bankomat.sprawdz_saldo()
    moj_bankomat.wyplata(200.00)
except (NieprawidlowaKwotaError, BrakSrodkowError) as e:
    print(f"Wystąpił błąd: {e}")

print("Drugi: Błąd 'Brak Środków' ")
try:

    moj_bankomat.wyplata(2000.00)
except BrakSrodkowError as e:
    print(f"Wyjątek Złapany (BrakSrodkowError) !")
    print(f"KOMUNIKAT: {e}")
    moj_bankomat.sprawdz_saldo()


print("Trzeci: Błąd 'Nieprawidłowa Kwota' ")
try:
    moj_bankomat.wyplata(-50.00)
except NieprawidlowaKwotaError as e:
    print(f"Wyjątek Złapany (NieprawidlowaKwotaError) !")
    print(f"KOMUNIKAT: {e}")

print("Czwarte: Błąd 'ValueError' (nie-liczba) ")
try:
    kwota_uzytkownika = "sto złotych"
    float(kwota_uzytkownika)
    moj_bankomat.wplata(float(kwota_uzytkownika))
except ValueError:
    print("Wyjątek Złapany (ValueError) !")
    print("Błąd: Wprowadzona wartość nie jest poprawną liczbą.")

moj_bankomat.sprawdz_saldo()
```

```python
class Macierz:

    def stworz_macierz_2x2():
        print("Wprowadzanie macierzy 2x2 ")

        try:
            w1_e1 = int(input("Podaj element (1, 1): "))
            w1_e2 = int(input("Podaj element (1, 2): "))
            w2_e1 = int(input("Podaj element (2, 1): "))
            w2_e2 = int(input("Podaj element (2, 2): "))
        except ValueError as e:

            print(f"FATALNY BŁĄD: Wprowadzono niepoprawną wartość! {e}")
            return [[0, 0], [0, 0]]

        macierz = [
            [w1_e1, w1_e2],
            [w2_e1, w2_e2]
        ]
        return macierz

    def wyswietl_macierz(self, macierz):
        print(f"| {macierz[0][0]} {macierz[0][1]} |")
        print(f"| {macierz[1][0]} {macierz[1][1]} |")

    def dodaj_macierze(self, macierz_a, macierz_b):
        print("Dodawanie macierzy")
        wynik = [
            [macierz_a[0][0] + macierz_b[0][0], macierz_a[0][1] + macierz_b[0][1]],
            [macierz_a[1][0] + macierz_b[1][0], macierz_a[1][1] + macierz_b[1][1]]
        ]
        return wynik

    def odejmij_macierze(self, macierz_a, macierz_b):
        print("Odejmowanie macierzy")
        wynik = [
            [macierz_a[0][0] - macierz_b[0][0], macierz_a[0][1] - macierz_b[0][1]],
            [macierz_a[1][0] - macierz_b[1][0], macierz_a[1][1] - macierz_b[1][1]]
        ]
        return wynik


macierz1 = Macierz.stworz_macierz_2x2()
print("Macierz 1:")
Macierz.wyswietl_macierz(macierz1)

macierz2 = Macierz.stworz_macierz_2x2()
print("Macierz 2:")
Macierz.wyswietl_macierz(macierz2)

suma = Macierz.dodaj_macierze(macierz1, macierz2)
print("Wynik dodawania:")
Macierz.wyswietl_macierz(suma)

roznica = Macierz.odejmij_macierze(macierz1, macierz2)
print("Wynik odejmowania:")
Macierz.wyswietl_macierz(roznica)
```

```python
class Ksiazka:
    def __init__(self, tytul, autor):
        self.tytul = tytul.strip().title()
        self.autor = autor.strip().title()
        self.dostepna = True

    def __str__(self):
        return f"'{self.tytul}' autorstwa {self.autor} ({'Dostępna' if self.dostepna else 'Wypożyczona'})"

class Czytelnik:
    def __init__(self, imie, nazwisko, numer_karty):
        self.imie = imie.strip().title()
        self.nazwisko = nazwisko.strip().title()
        self.numer_karty = numer_karty
        self.wypozyczone_ksiazki = []
    def __str__(self):
        return f"{self.imie} {self.nazwisko} (ID: {self.numer_karty})"

class Biblioteka:
    def __init__(self):
        self.lista_ksiazek = []
        self.lista_czytelnikow = []
        self.ostatni_numer_czytelnika = 1000

    def dodaj_ksiazke(self):
        tytul = input("Podaj tytuł książki: ")
        autor = input("Podaj autora książki: ")

        if not tytul or not autor:
            print(" Tytuł i autor nie mogą być puste.")
            return

        nowa_ksiazka = Ksiazka(tytul, autor)
        self.lista_ksiazek.append(nowa_ksiazka)
        print(f" Książka '{nowa_ksiazka.tytul}' została dodana do biblioteki.")

    def _generuj_numer_karty(self):
        self.ostatni_numer_czytelnika += 1
        return self.ostatni_numer_czytelnika

    def dodaj_czytelnika(self, imie, nazwisko):
        numer_karty = self._generuj_numer_karty()
        nowy_czytelnik = Czytelnik(imie, nazwisko, numer_karty)
        self.lista_czytelnikow.append(nowy_czytelnik)
        print(f" Czytelnik {nowy_czytelnik} został zarejestrowany.")

    def dodaj_czytelnika_z_inputem(self):
        imie = input("Podaj imię czytelnika: ")
        nazwisko = input("Podaj nazwisko czytelnika: ")

        if not imie or not nazwisko:
            print(" Imię i nazwisko nie mogą być puste.")
            return

        self.dodaj_czytelnika(imie, nazwisko)

        nowy_czytelnik = self.lista_czytelnikow[-1]
        print(f" Czytelnik {nowy_czytelnik.imie} {nowy_czytelnik.nazwisko} został dodany do systemu. Jego numer karty to: {nowy_czytelnik.numer_karty}")


    def wypozycz_ksiazke(self):
        tytul = input("Podaj tytuł książki do wypożyczenia: ")
        autor = input("Podaj autora książki do wypożyczenia: ")
        numer_karty_input = input("Podaj NUMER KARTY czytelnika: ")

        try:
            numer_karty = int(numer_karty_input)
        except ValueError:
            print(" Błąd: Numer karty musi być liczbą całkowitą.")
            return

        ksiazka_do_wypozyczenia = next((k for k in self.lista_ksiazek
                                        if k.tytul.lower() == tytul.lower() and k.autor.lower() == autor.lower()), None)

        if ksiazka_do_wypozyczenia is None:
            print(" Książka o podanym tytule i autorze nie została znaleziona.")
            return

        if not ksiazka_do_wypozyczenia.dostepna:
            print(" Ta książka jest już wypożyczona.")
            return


        czytelnik = next((c for c in self.lista_czytelnikow
                           if c.numer_karty == numer_karty), None)

        if czytelnik is None:
            print(f" Czytelnik o numerze karty '{numer_karty}' nie został znaleziony.")
            return


        ksiazka_do_wypozyczenia.dostepna = False
        czytelnik.wypozyczone_ksiazki.append(ksiazka_do_wypozyczenia)
        print(f" Pomyślnie wypożyczono '{ksiazka_do_wypozyczenia.tytul}' dla {czytelnik}.")

    def pokaz_status_ksiazek(self):
        if not self.lista_ksiazek:
            print("Brak książek w bibliotece.")
            return

        print("\n--- Zbiory Biblioteki ---")
        for k in self.lista_ksiazek:
            print(f"- {k}")
        print("-------------------------\n")

    def pokaz_wypozyczenia_czytelnika(self):
        numer_karty_input = input("Podaj NUMER KARTY czytelnika, aby zobaczyć wypożyczenia: ")

        try:
            numer_karty = int(numer_karty_input)
        except ValueError:
            print(" Błąd: Numer karty musi być liczbą całkowitą.")
            return

        czytelnik = next((c for c in self.lista_czytelnikow
                           if c.numer_karty == numer_karty), None)

        if czytelnik:
            if czytelnik.wypozyczone_ksiazki:
                print(f"\n--- Wypożyczenia {czytelnik.imie} {czytelnik.nazwisko} (ID: {czytelnik.numer_karty}) ---")
                for k in czytelnik.wypozyczone_ksiazki:
                    print(f"- '{k.tytul}' autorstwa {k.autor}")
                print("----------------------------------------------------------\n")
            else:
                print(f" Czytelnik {czytelnik.imie} {czytelnik.nazwisko} nie ma aktualnie żadnych wypożyczeń.")
        else:
            print(f" Czytelnik o numerze karty '{numer_karty}' nie został znaleziony.")

def menu():
    moja_biblioteka = Biblioteka()

    moja_biblioteka.dodaj_czytelnika("Barbara", "Kowalska")
    moja_biblioteka.dodaj_czytelnika("Jan", "Nowak")
    moja_biblioteka.lista_ksiazek.append(Ksiazka("Hobbit", "J.R.R. Tolkien"))
    moja_biblioteka.lista_ksiazek.append(Ksiazka("Zbrodnia i Kara", "Fiodor Dostojewski"))


    while True:
        print("\n=== System Biblioteczny (ID Użytkownika) ===")
        print("1. Dodaj nową książkę")
        print("2. Dodaj nowego czytelnika")
        print("3. Wyświetl wszystkie książki")
        print("4. Wypożycz książkę (wymaga NUMERU KARTY!)")
        print("5. Sprawdź wypożyczenia czytelnika (wymaga NUMERU KARTY!)")
        print("6. Zakończ")

        wybor = input("Wybierz opcję: ").strip()

        if wybor == '1':
            moja_biblioteka.dodaj_ksiazke()
        elif wybor == '2':
            moja_biblioteka.dodaj_czytelnika_z_inputem()
        elif wybor == '3':
            moja_biblioteka.pokaz_status_ksiazek()
        elif wybor == '4':
            moja_biblioteka.wypozycz_ksiazke()
        elif wybor == '5':
            moja_biblioteka.pokaz_wypozyczenia_czytelnika()
        elif wybor == '6':
            print("Dziękujemy za skorzystanie z systemu. Do widzenia!")
            break
        else:
            print(" Nieprawidłowy wybór. Spróbuj ponownie.")

if __name__ == "__main__":
    menu()
```
