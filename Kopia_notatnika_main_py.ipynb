wszystkie zwierzęta są istotami żywymi,
w zoo występują ssaki i ptaki,
każde zwierzę ma imię, gatunek i wiek,
każde zwierzę wie, w jakim pawilonie się znajduje,
każdy ssak ma informację o długości ciąży i liczbie młodych w miocie,
każdy ptak ma rozpiętość skrzydeł i informację, czy potrafi latać,
każdy pawilon ma swoją nazwę i listę zwierząt, które w nim mieszkają,
opis zwierzęcia ma zawierać wszystkie jego dane,
opis pawilonu ma zawierać nazwy wszystkich zwierząt i liczbę mieszkańców.
Dodatkowo:
Zoo potrafi wypisać ogólną liczbę zwierząt, średni wiek wszystkich zwierząt i liczbę gatunków.
Napisz program w Pythonie, który utworzy przykładowe zoo, niech trafią do niego zwierzęta. Program musi wypisać pełne obłożenie w zoo oraz statystyki.


```python
# Klasa bazowa reprezentująca Istotę Żywą (zwierzęta)
class IstotaŻywa:
    # Definicja klasy bazowej IstotaŻywa
    """Klasa bazowa dla wszystkich zwierząt w Zoo.""" # opis przeznaczenia klasy
    def __init__(self, imie, gatunek, wiek):
        # Metoda inicjalizująca (konstruktor) klasy, przyjmuje imię, gatunek i wiek
        # Sprawdzanie poprawności danych
        # Rozpoczęcie sekcji walidacji danych
        if not isinstance(wiek, int) or wiek < 0:
            # Sprawdzenie, czy wiek jest nieujemną liczbą całkowitą
            raise ValueError("Wiek musi być nieujemną liczbą całkowitą.")
            # Zgłoszenie błędu, jeśli wiek jest niepoprawny

        self.imie = imie
        # Przypisanie argumentu imie do atrybutu obiektu
        self.gatunek = gatunek
        # Przypisanie argumentu gatunek do atrybutu obiektu
        self.wiek = wiek
        # Przypisanie argumentu wiek do atrybutu obiektu
        self.pawilon = None  # Początkowo zwierzę nie jest przypisane do pawilonu
        # Inicjalizacja atrybutu pawilon na None (brak przypisanego pawilonu)

    def przypisz_pawilon(self, nazwa_pawilonu):
        # Metoda ustawiająca nazwę pawilonu
        """Ustawia nazwę pawilonu, w którym zwierzę się znajduje.""" # opis metody
        self.pawilon = nazwa_pawilonu
        # Przypisanie nazwy pawilonu do atrybutu pawilon

    def opis(self):
        # Metoda zwracająca podstawowy opis zwierzęcia
        """Zwraca podstawowy opis zwierzęcia, w tym informacje o pawilonie.""" # opis metody
        pawilon_info = self.pawilon if self.pawilon else "nieznany"
        # Ustalenie informacji o pawilonie: jeśli jest przypisany, użyj nazwy, w przeciwnym razie "nieznany"
        return (f"Imię: {self.imie}, Gatunek: {self.gatunek}, Wiek: {self.wiek} lat, Pawilon: {pawilon_info}")
        # Zwrócenie sformatowanego tekstu z podstawowymi danymi

# Klasa Ssak (dziedziczy po IstotaŻywa)
class Ssak(IstotaŻywa):
    # Definicja klasy Ssak, dziedziczącej po IstotaŻywa
    """Klasa dla ssaków, dodaje informacje o rozrodzie.""" # opis klasy
    def __init__(self, imie, gatunek, wiek, dlugosc_ciazy, liczba_mlodych):
        # Konstruktor klasy Ssak, przyjmuje dodatkowe parametry dla ssaków
        super().__init__(imie, gatunek, wiek)
        # Wywołanie konstruktora klasy nadrzędnej (IstotaŻywa) do ustawienia podstawowych atrybutów

        if not all(isinstance(val, int) and val >= 0 for val in [dlugosc_ciazy, liczba_mlodych]):
            # Sprawdzenie, czy dodatkowe atrybuty są nieujemnymi liczbami całkowitymi
            raise ValueError("Długość ciąży i liczba młodych muszą być nieujemnymi liczbami całkowitymi.")
            # Zgłoszenie błędu, jeśli walidacja się nie powiodła

        self.dlugosc_ciazy = dlugosc_ciazy  # w dniach/miesiącach
        # Przypisanie argumentu dlugosc_ciazy do atrybutu obiektu
        self.liczba_mlodych = liczba_mlodych
        # Przypisanie argumentu liczba_mlodych do atrybutu obiektu

    def opis(self):
        # Metoda opis, która nadpisuje (rozszerza) metodę z klasy nadrzędnej (polimorfizm)
        """Rozszerza opis o specyfikę ssaka.""" # opis metody
        podstawowy_opis = super().opis()
        # Wywołanie metody opis z klasy nadrzędnej (IstotaŻywa) i zapisanie jej wyniku
        return (f"{podstawowy_opis}, Długość ciąży: {self.dlugosc_ciazy}, Młodych w miocie: {self.liczba_mlodych}")
        # Zwrócenie rozszerzonego opisu o dane specyficzne dla ssaka

# Klasa Ptak (dziedziczy po IstotaŻywa)
class Ptak(IstotaŻywa):
    # Definicja klasy Ptak, dziedziczącej po IstotaŻywa
    """Klasa dla ptaków, dodaje informacje o lotności i rozpiętości skrzydeł."""# opis klasy
    def __init__(self, imie, gatunek, wiek, rozpietosc_skrzydel, potrafi_latac):
        # Konstruktor klasy Ptak, przyjmuje dodatkowe parametry dla ptaków
        super().__init__(imie, gatunek, wiek)
        # Wywołanie konstruktora klasy nadrzędnej (IstotaŻywa)

        if not isinstance(rozpietosc_skrzydel, (int, float)) or rozpietosc_skrzydel < 0:
            # Sprawdzenie, czy rozpiętość skrzydeł jest nieujemną liczbą (całkowitą lub zmiennoprzecinkową)
            raise ValueError("Rozpiętość skrzydeł musi być nieujemną liczbą.")
            # Zgłoszenie błędu, jeśli walidacja się nie powiodła
        if not isinstance(potrafi_latac, bool):
            # Sprawdzenie, czy potrafi_latac jest wartością logiczną (True/False)
            raise ValueError("Potrafi latać musi być wartością logiczną (True/False).")
            # Zgłoszenie błędu

        self.rozpietosc_skrzydel = rozpietosc_skrzydel  # w metrach
        # Przypisanie argumentu rozpietosc_skrzydel do atrybutu obiektu
        self.potrafi_latac = potrafi_latac
        # Przypisanie argumentu potrafi_latac do atrybutu obiektu

    def opis(self):
        # Metoda opis, która nadpisuje (rozszerza) metodę z klasy nadrzędnej
        """Rozszerza opis o specyfikę ptaka.""" # opis metody
        podstawowy_opis = super().opis()
        # Wywołanie metody opis z klasy nadrzędnej (IstotaŻywa)
        lotnosc = "Tak" if self.potrafi_latac else "Nie"
        # Ustalenie tekstu "Tak" lub "Nie" na podstawie wartości logicznej potrafi_latac
        return (f"{podstawowy_opis}, Rozpiętość skrzydeł: {self.rozpietosc_skrzydel:.2f} m, Potrafi latać: {lotnosc}")
        # Zwrócenie rozszerzonego opisu o dane specyficzne dla ptaka (formatowanie do 2 miejsc po przecinku)

# Klasa reprezentująca Pawilon w Zoo
class Pawilon:
    # Definicja klasy Pawilon
    """Klasa zarządzająca zwierzętami w danym pawilonie.""" # opis klasy
    def __init__(self, nazwa):
        # Konstruktor klasy Pawilon
        self.nazwa = nazwa
        # Przypisanie argumentu nazwa do atrybutu obiektu
        self.zwierzeta = []  # Lista przechowująca obiekty zwierząt
        # Inicjalizacja pustej listy do przechowywania obiektów IstotaŻywa

    def dodaj_zwierze(self, zwierze):
        # Metoda dodająca obiekt zwierzęcia do pawilonu
        """Dodaje zwierzę do listy i aktualizuje jego informację o pawilonie.""" # opis metody
        if isinstance(zwierze, IstotaŻywa):
            # Sprawdzenie, czy obiekt zwierze jest instancją klasy IstotaŻywa (lub jej podklasy)
            self.zwierzeta.append(zwierze)
            # Dodanie obiektu zwierzęcia do listy zwierząt w pawilonie
            zwierze.przypisz_pawilon(self.nazwa)
            # Ustawienie atrybutu pawilon w obiekcie zwierzęcia na nazwę tego pawilonu
        else:
            # Blok kodu wykonywany, jeśli obiekt nie jest Istotą Żywą
            print(f"Błąd: Obiekt {zwierze} nie jest poprawną Istotą Żywą.")
            # Wypisanie komunikatu o błędzie

    def opis(self):
        # Metoda zwracająca opis pawilonu
        """Zwraca opis pawilonu, w tym listę i liczbę zwierząt.""" # opis metody
        nazwy_zwierzat = [z.imie for z in self.zwierzeta]
        # Tworzenie listy imion wszystkich zwierząt w pawilonie (list comprehension)
        liczba_mieszkancow = len(self.zwierzeta)
        # Obliczenie liczby zwierząt w pawilonie

        return (
            # Zwrócenie sformatowanego opisu pawilonu, zawierającego:
            f"**Pawilon:** {self.nazwa}\n"
            # Nazwę pawilonu
            f"Liczba mieszkańców: {liczba_mieszkancow}\n"
            # Liczbę mieszkańców
            f"Nazwy zwierząt: {', '.join(nazwy_zwierzat) if nazwy_zwierzat else 'Brak'}"
            # Nazwy zwierząt (połączone przecinkami) lub "Brak", jeśli lista jest pusta
        )

# Klasa reprezentująca całe Zoo (centralny zarządca)
class Zoo:
    # Definicja głównej klasy Zoo
    """Główna klasa zarządzająca pawilonami i statystykami Zoo.""" # opis klasy
    def __init__(self, nazwa):
        # Konstruktor klasy Zoo
        self.nazwa = nazwa
        # Przypisanie argumentu nazwa do atrybutu obiektu
        self.pawilony = {}  # Słownik: {nazwa_pawilonu: obiekt_pawilonu}
        # Inicjalizacja pustego słownika do przechowywania obiektów Pawilon

    def dodaj_pawilon(self, pawilon):
        # Metoda dodająca obiekt Pawilon do Zoo
        """Dodaje obiekt pawilonu do Zoo."""# Docstring - opis metody
        if isinstance(pawilon, Pawilon):
            # Sprawdzenie, czy obiekt jest instancją klasy Pawilon
            self.pawilony[pawilon.nazwa] = pawilon
            # Dodanie obiektu Pawilon do słownika, używając nazwy pawilonu jako klucza
        else:
            # Blok kodu wykonywany, jeśli obiekt nie jest Pawilonem
            print(f"Błąd: Obiekt {pawilon} nie jest instancją klasy Pawilon.")
            # Wypisanie komunikatu o błędzie

    def _pobierz_wszystkie_zwierzeta(self):
        # Prywatna (konwencyjnie - nazwa zaczyna się od podkreślenia) metoda pomocnicza
        """Metoda pomocnicza zwracająca listę wszystkich zwierząt w Zoo."""
        # Docstring - opis metody
        wszystkie_zwierzeta = []
        # Inicjalizacja pustej listy
        for pawilon in self.pawilony.values():
            # Iteracja przez wszystkie obiekty Pawilon w słowniku pawilony
            wszystkie_zwierzeta.extend(pawilon.zwierzeta)
            # Dodanie (rozszerzenie) listy zwierząt z aktualnego pawilonu do listy wszystkich zwierząt
        return wszystkie_zwierzeta
        # Zwrócenie listy wszystkich zwierząt

    def oblicz_ogolna_liczbe_zwierzat(self):
        # Metoda do obliczania całkowitej liczby zwierząt
        """Zwraca ogólną liczbę zwierząt w Zoo."""# opis metody
        return len(self._pobierz_wszystkie_zwierzeta())
        # Zwrócenie długości listy wszystkich zwierząt

    def oblicz_sredni_wiek(self):
        # Metoda do obliczania średniego wieku zwierząt
        """Oblicza średni wiek wszystkich zwierząt."""# opis metody
        zwierzeta = self._pobierz_wszystkie_zwierzeta()
        # Pobranie listy wszystkich zwierząt
        if not zwierzeta:
            # Sprawdzenie, czy lista zwierząt jest pusta
            return 0
            # Zwrócenie 0, jeśli nie ma zwierząt
        suma_wieku = sum(z.wiek for z in zwierzeta)
        # Obliczenie sumy wieku wszystkich zwierząt (generator expression)
        return suma_wieku / len(zwierzeta)
        # Zwrócenie średniej arytmetycznej wieku

    def oblicz_liczbe_gatunkow(self):
        # Metoda do obliczania liczby unikalnych gatunków
        """Oblicza liczbę unikalnych gatunków w Zoo.""" # opis metody
        zwierzeta = self._pobierz_wszystkie_zwierzeta()
        # Pobranie listy wszystkich zwierząt
        gatunki = {z.gatunek for z in zwierzeta}
        # Utworzenie zbioru (set) unikalnych nazw gatunków (set comprehension)
        return len(gatunki)
        # Zwrócenie liczby elementów w zbiorze (czyli liczby unikalnych gatunków)

    def opis_zoo(self):
        # Metoda do wypisania pełnego opisu i statystyk Zoo
        """Wypisuje pełny opis Zoo, pawilonów i statystyk."""# opis metody
        print("=" * 70)
        # Wypisanie linii rozdzielającej
        print(f"**Witaj w ZOO:** {self.nazwa.upper()} ")
        # Wypisanie powitania z nazwą Zoo (dużymi literami)
        print("=" * 70)
        # Wypisanie linii rozdzielającej

        # 1. Opis pawilonów
        # Sekcja opisu pawilonów
        print("\n### Pawilony i ich mieszkańcy:")
        # Nagłówek sekcji
        if not self.pawilony:
            # Sprawdzenie, czy słownik pawilonów jest pusty
            print("Zoo nie posiada jeszcze żadnych pawilonów.")
            # Komunikat o braku pawilonów
        else:
            # Jeśli są pawilony
            for pawilon in self.pawilony.values():
                # Iteracja przez wszystkie obiekty Pawilon
                print("-" * 30)
                # Wypisanie linii rozdzielającej opisy pawilonów
                print(pawilon.opis())
                # Wywołanie metody opis() dla każdego obiektu Pawilon

        # 2. Szczegółowy opis zwierząt (polimorfizm w akcji!)
        # Sekcja szczegółowego spisu zwierząt
        print("\n### Szczegółowy spis zwierząt:")
        # Nagłówek sekcji
        wszystkie_zwierzeta = self._pobierz_wszystkie_zwierzeta()
        # Ponowne pobranie listy wszystkich zwierząt
        for z in wszystkie_zwierzeta:
            # Iteracja przez wszystkie obiekty zwierząt
            print(f"  - {z.opis()}")
            # Wypisanie szczegółowego opisu każdego zwierzęcia (wykorzystanie polimorfizmu - wywołuje opis z odpowiedniej podklasy)

        # 3. Statystyki Zoo
        # Sekcja statystyk
        print("\n### Statystyki Zoo:")
        # Nagłówek sekcji
        print(f"Ogólna liczba zwierząt: **{self.oblicz_ogolna_liczbe_zwierzat()}**")
        # Wypisanie całkowitej liczby zwierząt
        print(f"Średni wiek wszystkich zwierząt: **{self.oblicz_sredni_wiek():.2f} lat**")
        # Wypisanie średniego wieku (sformatowane do 2 miejsc po przecinku)
        print(f"Liczba unikalnych gatunków: **{self.oblicz_liczbe_gatunkow()}**")
        # Wypisanie liczby unikalnych gatunków
        print("=" * 70)
        # Wypisanie linii kończącej opis

# --- Użycie i testowanie ---

# 1. Tworzenie obiektów zwierząt
# Sekcja testowa: Tworzenie obiektów zwierząt
# Ssak: (imię, gatunek, wiek, długość ciąży, liczba młodych)
# Komentarz objaśniający kolejność argumentów
ssak_leo = Ssak("Simba", "Lew afrykański", 10, 110, 3)
# Utworzenie obiektu Ssak (Lew)
ssak_gina = Ssak("Nadira", "Żyrafa", 5, 457, 1)
# Utworzenie obiektu Ssak (Żyrafa)

# Ptak: (imię, gatunek, wiek, rozpiętość skrzydeł, potrafi latać)
# Komentarz objaśniający kolejność argumentów
ptak_eagles = Ptak("Ares", "Orzeł bielik", 7, 2.4, True)
# Utworzenie obiektu Ptak (Orzeł)
ptak_ping = Ptak("Fraczek", "Pingwin cesarski", 3, 0.9, False)
# Utworzenie obiektu Ptak (Pingwin)

# 2. Tworzenie pawilonów
# Sekcja testowa: Tworzenie obiektów Pawilon
pawilon_sawanna = Pawilon("Sawanna")
# Utworzenie obiektu Pawilon o nazwie "Sawanna"
pawilon_woliera = Pawilon("Woliera i Wody")
# Utworzenie obiektu Pawilon o nazwie "Woliera i Wody"

# 3. Dodawanie zwierząt do pawilonów (automatycznie przypisuje pawilon zwierzęciu)
# Sekcja testowa: Dodawanie zwierząt do pawilonów
pawilon_sawanna.dodaj_zwierze(ssak_leo)
# Dodanie Lwa do Pawilonu "Sawanna"
pawilon_sawanna.dodaj_zwierze(ssak_gina)
# Dodanie Żyrafy do Pawilonu "Sawanna"
pawilon_woliera.dodaj_zwierze(ptak_eagles)
# Dodanie Orła do Pawilonu "Woliera i Wody"
pawilon_woliera.dodaj_zwierze(ptak_ping)
# Dodanie Pingwina do Pawilonu "Woliera i Wody"

# 4. Tworzenie obiektu Zoo
# Sekcja testowa: Tworzenie obiektu Zoo
moje_zoo = Zoo("Zoo Wrocław")
# Utworzenie głównego obiektu Zoo

# 5. Dodawanie pawilonów do Zoo
# Sekcja testowa: Dodawanie pawilonów do Zoo
moje_zoo.dodaj_pawilon(pawilon_sawanna)
# Dodanie Pawilonu "Sawanna" do Zoo
moje_zoo.dodaj_pawilon(pawilon_woliera)
# Dodanie Pawilonu "Woliera i Wody" do Zoo

# 6. Wypisanie opisu całego ekosystemu Zoo
# Sekcja testowa: Wyświetlenie wyników
moje_zoo.opis_zoo()
# Wywołanie metody opis_zoo() w celu wyświetlenia wszystkich informacji i statystyk
```
